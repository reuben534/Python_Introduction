```python
import pandas as pd #data manipulation
import numpy as np #numeric/specific computation
import matplotlib.pyplot as plt  #data visualisation
import seaborn as sns  #data visualisation
```

```python
df = pd.read_excel('/content/Sample - Superstore.xlsx')
```

```python
df.head()
```

```python
df.columns
```

```python
df.info
```

```python
# Bar Plot
#P.S. 1 What is the total sales generated by each product category
res = df.groupby("Category")['Sales'].sum().reset_index().sort_values('Sales', ascending = False)
print(res)
plt.figure(figsize= (8,3))
sns.barplot(x = res['Category'], y = res['Sales'], color = 'green')
```

```python
# Count Plot - count of each distinct catergory of a categorical column as a bar plot
plt.figure(figsize= (8,3))
sns.countplot(x = df['Category'], color= 'green')
```

```python
# Find average sales for each region
avg_sales = df.groupby('Region')['Sales'].mean().reset_index()

print(avg_sales)

# Visualize
plt.figure(figsize=(8,3))
sns.barplot(data=avg_sales, x='Region', y='Sales', palette='viridis')

plt.title('Average Sales Generated by Each Region')
plt.xlabel('Region')
plt.ylabel('Average Sales')
plt.show()
```

```python
segment_count = df.groupby('Segment').size().reset_index(name='Count')

print(segment_count)
plt.figure(figsize=(8,3))
sns.countplot(x='Segment', data=df, palette='Set2')

plt.title('Count of Products from Each Segment')
plt.xlabel('Segment')
plt.ylabel('Count')
plt.show()
```
