#NumPy
Introduction to NumPy Arrays

NumPy (Numerical Python) is a powerful Python library used for numerical computations, mathematical operations, and working with arrays efficiently.

```python
#Import NumPy
import numpy as np
```

```python
#Check the Type of NumPy
type(np)
```

```python
#Create Your First NumPy Array
a1 = np.array([1, 2, 3, 4])
a1
```

#Array Attributes

```python
#Shape
a1.shape
```

```python
#Size
a1.size
```

#Data Type

```python
a1.dtype
```

```python
#Number of Dimensions
a1.ndim
```

#Arrays with Different Data Types

```python
#Mixed Integers and Floats
a2 = np.array([20, 119, 77.7, 10.09, 11.11, 10000, 700])
a2
```

#Arithmetic Operations

```python
#Create Arrays
a3 = np.array([10, 20, 30])

a4 = np.array([1, 2, 3])
```

```python
#Addition
a3 + a4
```

```python
#Subtraction
a3 - a4
```

```python
#Multiplication
a3 * a4
```

```python
#Division
a3 / a4
```

```python
#Modulus
a3 % a4
```

```python
#Exponent
a3 ** a4
```

```python
#Floor Division
a3 // a4
```

```python
#Operations with Scalars
a3 * 100
```

```python
#Addition
a3 + 10
```

```python
#Subtraction
a3 - 10
```

```python
#Division
a3 / 10
```

#Statistical Operations

```python
#Mean (Average)
a3.mean()
```

```python
#Method 2
np.mean(a3)
```

```python
#Median
np.median(a3)
```

```python
#Maximum Value
#Method 1
np.max(a3)
```

```python
#Method 2
a3.max()
```

```python
#Minimum Value
np.min(a3)
```

```python
#Variance
np.var(a3)
```

```python
#Standard Deviation
np.std(a3)
```

```python
#Position of Maximum Value
a3.argmax()
```

```python
#Position of Minimum Value
a3.argmin()
```

```python
np.mean(a3)      # Mean
np.median(a3)    # Median
np.min(a3)       # Minimum
np.max(a3)       # Maximum
np.var(a3)       # Variance
np.std(a3)       # Standard Deviation
```
