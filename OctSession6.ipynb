#Practice Task Solution

```python
# Create a DataFrame
students_df = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104, 105, 106],
    "Marks": [85, 90, np.nan, 78, 95, np.nan],
    "Attendance": [90, 85, 88, 92, 95, 80],
    "Age": [20, 21, 22, np.nan, 21, 20]
})

students_df
```

```python
#Information About the DataFrame
students_df.info()
```

```python
#Statistical Summary
students_df.describe()
```

```python
#View First 5 Records
students_df.head()
```

```python
#View Last 5 Records
students_df.tail()
```

```python
#Count Missing Values
students_df.isnull().sum()
```

#Sort DataFrame by Marks

```python
#Ascending Order
students_df.sort_values(by="Marks")
```

```python
#Descending Order
students_df.sort_values(by="Marks", ascending=False)
```

```python
#Frequency Count
students_df["Age"].value_counts()
```

```python
#Unique Values
students_df["Attendance"].unique()
```

```python
#Count Unique Values
students_df["Attendance"].nunique()
```

#Statistical Operations

```python
#Mean
students_df["Marks"].mean()
```

```python
#Median
students_df["Marks"].median()
```

```python
#Mode
students_df["Age"].mode()
```

```python
#Minimum Value
students_df["Marks"].min()
```

```python
#Maximum Value
students_df["Marks"].max()
```

```python
#Variance
students_df["Marks"].var()
```

```python
#Standard Deviation
students_df["Marks"].std()
```

```python
#Perform All Statistical Operations at Once
print("Mean:", students_df["Marks"].mean())
print("Median:", students_df["Marks"].median())
print("Mode:")
print(students_df["Marks"].mode())

print("Minimum:", students_df["Marks"].min())
print("Maximum:", students_df["Marks"].max())

print("Variance:", students_df["Marks"].var())
print("Standard Deviation:", students_df["Marks"].std())
```

```python
#Handle Missing Values
#Fill Missing Values with Mean

students_df["Marks"] = students_df["Marks"].fillna(
students_df["Marks"].mean()
)
students_df["Age"] = students_df["Age"].fillna(
students_df["Age"].median()
)
students_df
```

```python
#Remove Missing Values
students_cleaned = students_df.dropna()
students_cleaned
```
