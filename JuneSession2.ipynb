###This program accepts marks for three modules and calculates the total and average.

```python
m1 = int(input("Module 1 marks: "))
m2 = int(input("Module 2 marks: "))
m3 = int(input("Module 3 marks: "))

print(m1, m2, m3)

total = m1 + m2 + m3
print(total)

average_marks = total / 3
print(average_marks)
```

##List Derived Data Type

```python
#A list is a collection of multiple values stored in a single variable.
marks = [5, 6, 8, 9, 7, 10, 10, 7, 6, 6]
print("Marks scored:", marks)
```

```python
#Check List Type
type(marks)
```

##Indexing in Lists
.Python starts counting positions from 0.

```python
#First Element
marks[0]
```

```python
#Fourth Element
marks[3]
```

##Negative Indexing

```python
#Negative indexing starts from the end of the list.
marks[-4]
```

```python
#List Reference
marks
```

##Slicing

```python
"""Slicing retrieves a portion of a list.
Explanation:
Starts at index 2
Ends before index 7
"""
marks[2:7]
```

##List Functions

```python
#The append() method adds a value to the end of the list
marks.append(50)
print(marks)
```

##insert()

```python
"""The insert() method adds a value at a specified position.
Note: If the index exceeds the list length,
Python inserts the value at the end.
"""
marks.insert(100, 6)
print(marks)
```

```python
"""6 is the position (index)
100 is the value inserted
"""
marks.insert(6, 100)
print(marks)
```

```python
marks = [80, 75, 90]
marks.append(85)
marks.insert(1, 95)
print(marks)
```
