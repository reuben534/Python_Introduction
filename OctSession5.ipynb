#Creating DataFrames

```python
#Create a DataFrame Using a List
learners = [
    [101, "Getrude", "Data Science"],
    [102, "Bongani", "Power BI"],
    [103, "Keren", "GENAI"]
]

df_1 = pd.DataFrame(
    learners,
    columns=["ID", "Names", "Course"]
)

df_1
```

```python
#Create a DataFrame Using a Dictionary
learners_dict = {
    "ID": [101, 102, 103],
    "Names": ["Getrude", "Bongani", "Keren"],
    "Course": ["Data Science", "Power BI", "GENAI"]
}

learners_df = pd.DataFrame(learners_dict)

learners_df
```

```python
#. Create a Series Using a List
s1 = pd.Series(
    ["South Africa", "India", "Nigeria", "Uganda"]
)

s1
```

```python
#Create a Series Using a List with Custom Indexes
s2 = pd.Series(
    [1000, 2000, 3000, 4000, 5000.999],
    index=[
        "Cape Town",
        "Durban",
        "Pretoria",
        "Mumbai",
        "Kisumu"
    ]
)

s2
```
