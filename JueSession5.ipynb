##Creating a For Loop

```python
#Print "Hello" Five Times
for i in range(5):
    print("Hello!!")
```

```python
for value in range(5):
    print("Hello")
```

```python
#Printing Numbers Using Loops
for i in range(5):
    print(i)
```

```python
#Printing on the Same Line
for i in range(5):
    print(i, end="   ")
```

```python
#Task 4: Print Numbers From 20 to 30
for i in range(20, 31):
    print(i, end=" ")
```

```python
#Task 5: Print "Regenesys" Three Times
for i in range(3):
    print("Regenesys")
```

```python
#Task 6: Print Table of 7
for i in range(7, 71, 7):
    print(i, end=" ")
```

```python
#Write a for loop to display the table of 11.
for i in range(11, 111, 11):
    print(i, end=" ")
```

##Reverse Order Loops

```python
#Print Numbers 5 to 1
for i in range(5, 0, -1):
    print(i)
```

```python
#Table of 7 in Reverse Order
for i in range(70, 6, -7):
    print(i, end="  ")
```

```python
#Table of 2 in Reverse Order
for i in range(20, 1, -2):
    print(i, end="   ")
```

##Traversing a List Using a Loop

```python
#Lists can be processed one element at a time using a loop.
learners = ["Keren", "George", "Bongani", "Jerry", "Shadi"]

for learner in learners:
    print(f"Hello {learner} !!")
```

##Using f-Strings

```python
#An f-string allows variables to be inserted into text.
age = 50
name = "John"
print(f"My name is {name} and my age is {age}")
```

```python
#Customized Welcome Message
learners = ["Keren", "George", "Bongani", "Jerry", "Shadi"]
for learner in learners:
    print(f"Welcome to Regenesys, {learner}")
```

```python
#Task 7: Create Your Own List and Print a Customized Statement
friends = ["Thato", "Ryan", "Walter", "Elias"]
for friend in friends:
    print(f"{friend} is learning Python programming.")
```

##While Loop

```python
#A while loop repeats a task as long as a condition remains true.
count = 1
while count <= 5:
    print(count)
    count += 1
```
