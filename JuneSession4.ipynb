##1. If - Elif - Else Statement

```python
#Problem Statement 1
#Write a program to check whether a given number is positive, negative, or zero.
num = -7

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

##2. Nested If Statement

```python
#Problem Statement 2
#Write a program to check whether a person is eligible to vote.
age = 17

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

```python
#input example age and nationality
age = int(input("Enter your age: "))
nationality = input("Enter your nationality: ")

# We use the 'in' operator to check if the input nationality is in our list of allowed nationalities
allowed_nationalities = ["Indian", "White", "Black", "other"]

if age >= 18:
    if nationality in allowed_nationalities:
        print("Eligible to vote")
    else:
        print("Not eligible to vote due to nationality restrictions")
else:
    print("Not eligible to vote due to age")
```

#Task 2: Salary Increment Based on Experience

```python
#Problem Statement
#Increase salary according to years of experience.
#Experience	Increment
#0 - 5 years	5%
#More than 5 years and up to 10 years	10%
#More than 10 years	15%
salary = float(input("Enter current salary: "))
experience = int(input("Enter years of experience: "))

if experience <= 5:
    increment = salary * 0.05

elif experience <= 10:
    increment = salary * 0.10

else:
    increment = salary * 0.15

new_salary = salary + increment

print("Increment:", increment)
print("New Salary:", new_salary)
```

```python
#3. Check Whether a Year is a Leap Year
year = int(input("Enter a year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

##4. Calculate Discount Based on Purchase Amount


```python
#Purchase Amount	Discount
#Less than 1000	5%
#1000 - 5000	10%
#Above 5000	15%

amount = float(input("Enter purchase amount: "))

if amount < 1000:
    discount = amount * 0.05

elif amount <= 5000:
    discount = amount * 0.10

else:
    discount = amount * 0.15

final_amount = amount - discount

print("Discount =", discount)
print("Amount to Pay =", final_amount)
```

##5. Determine Whether a Person Qualifies for a Driver's License


```python
#Requirements:
#Must be 18 years or older.
#Must pass the eye test.
age = int(input("Enter your age: "))
eye_test = input("Did you pass the eye test? (Yes/No): ")

if age >= 18:
    if eye_test == "Yes":
        print("Qualified for Driver's License")
    else:
        print("Failed Eye Test")

else:
    print("Under Age - Not Qualified")
```

##Write a program that assigns grades using the following criteria:

```python
#Marks	Grade
#80 - 100	A
#70 - 79	B
#60 - 69	C
#50 - 59	D
#Below 50	F
marks = float(input("Enter marks: "))

if marks >= 80:
    print("Grade A")

elif marks >= 70:
    print("Grade B")

elif marks >= 60:
    print("Grade C")

elif marks >= 50:
    print("Grade D")

else:
    print("Grade F")
```
