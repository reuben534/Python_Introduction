```python
# Import pandas for data analysis and manipulation
import pandas as pd

# Import NumPy for numerical operations
import numpy as np

# Load the Superstore Excel dataset into a DataFrame
df = pd.read_excel('/content/Sample - Superstore.xlsx')

# Display the first 10 rows of the dataset
df.head(10)
```

```python
# Get the total number of elements (rows × columns) in the DataFrame
df.size
```

```python
# Display information about the DataFrame, including columns, data types, non-null counts, and memory usage
df.info()
```

```python
# Generate descriptive statistics for all numerical columns in the DataFrame
df.describe()
```

```python
# Display the number of rows and columns in the DataFrame
df.shape
```

```python
# Display the data type of each column in the DataFrame
df.dtypes
```

```python
# Check for missing (null) values in each column of the DataFrame
df.isnull().sum()
```

```python
# Remove all rows that contain missing (null) values and store the cleaned data in a new DataFrame
df_new = df.dropna()
```

```python
# Display the number of rows and columns in the cleaned DataFrame after removing missing values
df_new.shape
```

```python
# Calculate the total profit earned by the Superstore
total_profit = df["Profit"].sum()

# Display the total profit
print("Total Profit Generated by the Superstore:", total_profit)
```

```python
# Count the total number of orders in the dataset
total_orders = df["Order ID"].count()

# Display the total number of orders placed
print(f"Total Orders Placed: {total_orders}")
```

```python
# Calculate the number of unique customers in the Superstore dataset
total_unique_customers = df["Customer ID"].nunique()

# Display the total number of unique customers
print(f"Total Unique Customers: {total_unique_customers}")
```

```python
# Count the number of products/orders in each category
df["Category"].value_counts()
```

```python
# Count the occurrences of each category and convert the result into a DataFrame
df["Category"].value_counts().reset_index()
```

```python
# Count the number of records in each region
df["Region"].value_counts()
```
