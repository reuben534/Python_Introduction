##Non-Primitive Data Types in Python

##1. Lists

```python
"""
A list is a data structure used to store a collection of elements.
Characteristics
.Elements are separated by commas and enclosed in square brackets [].
.Lists are ordered (the position of items matters).
.Lists are mutable (can be modified).
.Lists support indexing.
.Indexing starts from 0 and ends at N-1, where N is the number of elements.
"""
dishes = ["rice", "chicken", "lamb curry", "ugali"]
dishes[0]
```

##Deleting Elements from a List


```python
#Removes an item by its value.
dishes.remove("chicken")
```

```python
#or print(dishes)
dishes
```

##2. pop()

```python
#Removes an item using its index and returns the removed item.
dishes.pop(1)
```

```python
print(dishes)
```

##3. extend()

```python
#Adds multiple items to a list.
dishes.extend(["tea", "coffee"])
```

```python
print(dishes)
```

##4. del Keyword

```python
#Removes an item at a specific index.
del dishes[3]
```

```python
print(dishes)
```

```python
del dishes[3]
```

```python
print(dishes)
```

##5. clear()

```python
#Removes all items from a list.
dishes.clear()
```

```python
print(dishes)
```

##6. Delete the Entire List

```python
del dishes
```

```python
dishes
```

##3. Dictionary
#####A dictionary stores data as key-value pairs.
#####Real-Life Analogy
#####A traditional dictionary contains:
#####Word → Meaning
####Similarly, a Python dictionary stores:
#####Key → Value
#####Characteristics
#####Enclosed in curly braces {}.
#####Consists of key-value pairs.
#####Mutable (can be changed).
#####Keys must be unique.
#####Values can be duplicated.
#####Items are stored as key-value pairs.

```python
learners = {
    101: "Bongani",
    102: "John",
    103: "Keren",
    104: "Shadi"
}
```

```python
learners[102]
```

```python
learners
```

```python
learn = {
    300: "John",
    1000: "Bongani",
    -1: "Keren"
}
```

```python
learn
```

```python
learn[-1]
```

##Population Data

```python
population = {
    "Kampala": 20000000,
    "Lagos": 25000000,
    "Johannesburg": 15000000,
    "Pretoria": 1500000,
    "Cape Town": 4000000
}
```

```python
population["Lagos"]
```

```python
len(population)
```

##Deleting Items from a Dictionary

```python
#1. pop()
#Removes an item using its key.
population.pop("Lagos")
```

```python
population
```

```python
#2. popitem()
#Removes and returns the last inserted item.
population.popitem()
```

##3. del Keyword

```python
del population["Johannesburg"]
```

```python
population
```

##4. clear()

```python
#Removes all items from the dictionary.
population.clear()
```

```python
population
```

##Adding Items to a Dictionary

```python
#Using update()
population.update({"Lagos": 15000000})
```

```python
population
```

```python
#Adding multiple items:
population.update({
    "Kampala": 10000000,
    "Pretoria": 1500000
})
```

```python
population
```

#Using Assignment Operator (=)

```python
population["Cape Town"] = 15000000
```

```python
population
```
